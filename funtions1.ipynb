```python
def name():
    print("Hello, World!")

name()
```

```python
def add(a,b):
    return a+b
print(add(9,8))
```

```python
def user(name):
    return f"Hello, {name}!"
print(user("Ravi"))
```

```python
def student(name,marks):
    print(f'{name} scored {marks} marks')

student('nani',89)
```

```python
def info(name,age):
    print('name:',name)
    print('age:',age)

info(age=90,name='somu')
```

```python
def info(city='Hyderabad',country='India'):
    print('city:',city)
    print('country:',country)

info(city='washington',country='USA')
```

```python
def num(*numbers):
    print('numbers:',numbers)

num(1,2,3,4,6)
```

```python
def sub(*args):
    return sum(args)

print(sub(10,20,30))
```

```python
def name(**info):
    print('info:',info)

name(name='nani',age=21,occ='student',city='Hyderabad')
```

```python
def marks(**student)    :
    print('student:',student)

marks(name='rahu',age=15,marks=89)
```

```python
add=lambda a,b:a+b
print(add(1,2))
```

```python
sqr=lambda x:x**2
print(sqr(5))
```

```python
great=lambda x,y:x if x>y else y
print(great(10,20))
```

```python
even_odd=lambda x:x%2==0
print(even_odd(10))
```

```python
even_odd=lambda x:'even' if x%2==0 else 'odd'
print(even_odd(11))
```

```python
import math
print(math.sqrt(2))
print(math.pow(2,3))
print(math.ceil(2.3))
print(math.floor(2.7))
print(math.factorial(8))
```
