```python
import pandas as pd

data = {
    'name': ['Somu', 'Ravi', 'Harry', 'Appy', 'Kiran', 'Raju'],
    'marks': [85, 78, 92, 66, 74, 88]
}

df = pd.DataFrame(data)

# Top performer
top = df[df['marks'] == df['marks'].max()]
print("Top performer:")
print(top)

# Second highest
second = df.sort_values(by='marks', ascending=False).iloc[1]
print("\nSecond highest:")
print(second)
```
