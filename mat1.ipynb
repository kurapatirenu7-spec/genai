```python
import matplotlib.pyplot as plt
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]
plt.xlabel('term1')
plt.ylabel('term2')
plt.title('the x and y')
plt.plot(x,y)
plt.show()
```

```python


import matplotlib.pyplot as plt
x = [1,2,3,4,5]
y1 = [1,2,3,4,5]
y2 = [1,4,9,16,25]
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.title('multiple lines')
plt.plot(x,y1,label='linear')
plt.plot(x,y2,label='square')
plt.legend()
plt.show()
```

```python
x=[12, 15, 14, 10, 9, 8, 7, 15, 16, 18, 20]
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title('histogram')
plt.hist(x,bins=5,color='blue',edgecolor='pink')
plt.show()
```

```python
x = [30, 25, 20, 25]
labels = ['Pizza', 'Burger', 'Pasta', 'Salad']
plt.title('food')
plt.pie(x,labels=labels,autopct='%1.1f%%',startangle=90,shadow=True)
plt.axis('equal')
plt.show()
```

```python
students = ['A', 'B', 'C', 'D', 'E']

math_marks = [80, 70, 90, 60, 85]
science_marks = [75, 85, 80, 70, 90]
plt.xlabel('math_marks')
plt.ylabel('science_marks')
plt.title('marks')
plt.plot(students,math_marks,label='mathmarks',marker='o')
plt.plot(students,science_marks,label='sciencemarks',marker='o')
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']

temperature = [32, 34, 31, 35, 36, 33, 30]
plt.xlabel('days')
plt.ylabel('temperature')
plt.title('temperature of some cities')
plt.plot(days,temperature,marker='o')
plt.grid(True)
plt.show()
```

```python
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri']

temperature = [30, 32, 29, 35, 33]

sales = [100, 120, 90, 150, 130]
plt.subplot(1, 2, 1)
plt.title('temperature')
plt.xlabel('Days')
plt.ylabel('Temperature (°C)')
plt.plot(days,temperature,label='temperature',marker='o')
plt.legend()
plt.subplot(1, 2, 2)
plt.title('Sales')
plt.xlabel('Days')
plt.ylabel('Sales')
plt.bar(days,sales,color='pink')
plt.title('sales')
plt.show()
```
