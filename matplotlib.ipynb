```python
import matplotlib.pyplot as plt
student=['Alice','Bob','Charlie','David','Eve']
grades=[85,90,78,92,88]
plt.title('Student Grades')
plt.xlabel('Students')
plt.ylabel('Grades')
plt.bar(student, grades)
plt.show()
```

```python
patient=['somu','sana','nani','siri','soma']
disesase=['fever','cold','cough','headache','stomachache']
plt.title('Patient Diseases')
plt.xlabel('Patients')
plt.ylabel('Diseases')
plt.plot(patient, disesase)
plt.show()
```

```python
branch=['Computer Science','Mechanical','Electrical','Civil','Chemical']
students=[120,100,80,60,40]
plt.title('Students in Different Branches')
plt.xlabel('Branches')
plt.ylabel('Number of Students')
plt.scatter(branch, students)
plt.show()
```

```python
kdrama=['Action','Comedy','Drama','Horror','Sci-Fi']
ratings=[4.5,4.0,4.8,3.5,4.2]
plt.title('K-Drama Ratings')
plt.xlabel('Genres')
plt.ylabel('Ratings')
plt.figure(figsize=(6, 6))
plt.pie(ratings, labels=kdrama, autopct='%1.1f%%')
plt.show()
```

```python
movies=['Inception','The Dark Knight','Interstellar','Django Unchained','The Matrix']
box_office=[829,1005,677,425,463]
plt.title('Box Office Collection')
plt.xlabel('Movies')
plt.ylabel('Box Office (in millions)')
plt.hist(box_office, bins=5, edgecolor='black')
plt.show()
```
