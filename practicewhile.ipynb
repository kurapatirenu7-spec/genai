```python
def palin(s):
    rev=''
    for i in s:
        rev=i+rev
    if s==rev:
        return 'palindrome'
    return 'not a palindrome'
    

print(palin('cool'))
```

```python
def character(n):
    vowels='aeiouAEIOU'
    count=0
    for i in n:
        if i in vowels:
            count+=1
    return count

print(character('renuka'))
```

```python
def add(a):
    count=0
    for i in a:
        count+=i
    return count
a=[10,20,30,40,50]
print(add(a))
```

```python
def count_consonants(n):
    vowels='aeiouAEIOu'
    count=0
    for i in n:
        if i.isalpha() and i not in vowels:
            count+=1
    return count
print(count_consonants('hello'))
```

```python
def maximum(a):
    large=a[0]
    for i in a:
        if i>large:
            large=i
    return large
a=[10,24,56,7,8]
print(maximum(a))
```

```python
def fibo(n):
    a,b=0,1
    for i in range(n):
        a,b=b,a+b
    return a
print(fibo(5))
```
