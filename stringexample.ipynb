```python
n=input()
a=e=i=o=u=0
for ch in n.lower():
    if ch=='a':
        a+=1
    elif ch=='e':
        e+=1
    elif ch=='i':
        i+=1
    elif ch=='o':
        o+=1
    elif ch=='u':
        u+=1
print('a:',a)
print('e:',e)
print('i:',i)
print('o:',o)
print('u:',u)
```

```python


n=input()
count1=0
count2=0
for ch in n:
    if ch.isupper():
        count1+=1
    elif ch.islower():
        count2+=1
print(f'{n}={count1},{count2}')
```
